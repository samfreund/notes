```python
import json
import os
import pandas as pd

def readEmail(filename):
    with open(filename, 'r') as f:
        email_data = json.load(f)
    return email_data

emails = []

for file in os.listdir('email_json'):
    if file.endswith('.json'):
        email_data = readEmail(os.path.join('email_json', file))
        emails.append(email_data)

df = pd.DataFrame(emails)

df['body'] = df['body'].str.replace('\n', '')
df['body'] = df['body'].str.replace('\'', '')

print(df.head())
```

The column names are `to_address`, `from_address`, `subject`, and `body`. #TODO datatypes

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(binary=True, min_df=10)
body_feature_matrix = vectorizer.fit_transform(df['body'].fillna(''))

print('Feature matrix shape:', body_feature_matrix.shape)
print('Sample feature names:', vectorizer.get_feature_names_out()[:20])
```
